```python

import pandas as pd
from pgmpy.models import DiscreteBayesianNetwork
from pgmpy.estimators import BayesianEstimator
from pgmpy.inference import VariableElimination

# 1. Load Dataset
data = pd.read_csv("heart.csv")

print("Heart Disease Dataset:")
print(data.head())

# 2. Select Important Attributes
data = data[['age', 'sex', 'cp', 'exang', 'ca', 'thal', 'target']]

# 3. Convert Age into Categories
data['age'] = pd.cut(
    data['age'],
    bins=[0, 40, 55, 100],
    labels=[0, 1, 2],
    include_lowest=True
).astype(int)

# 4. Create Bayesian Network
model = DiscreteBayesianNetwork([
    ('age', 'target'),
    ('sex', 'target'),
    ('cp', 'target'),
    ('exang', 'target'),
    ('ca', 'target'),
    ('thal', 'target')
])

# 5. Train Bayesian Network
model.fit(
    data,
    estimator=BayesianEstimator,
    prior_type='BDeu'
)

# 6. Check Model
print("\nIs Model Valid?", model.check_model())

# 7. Perform Inference
inference = VariableElimination(model)

result = inference.query(
    variables=['target'],
    evidence={
        'age': 1,
        'sex': 1,
        'cp': 2,
        'exang': 0,
        'ca': 0,
        'thal': 2
    }
)

# 8. Display Prediction
print("\nPrediction Result:")
print(result)

# 9. Display Heart Disease Probability
probability = result.values[1] * 100

if probability >= 50:
    print("\nHeart Disease: Present")
else:
    print("\nHeart Disease: Not Present")

print("Probability:", round(probability, 2), "%")
```
