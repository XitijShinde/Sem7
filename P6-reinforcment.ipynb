```python

import numpy as np
import random

# Maze: 0 = Free, 1 = Wall
maze = [
    [0, 0, 0, 0],
    [1, 1, 0, 1],
    [0, 0, 0, 0],
    [0, 1, 1, 0]
]

start = (0, 0)
goal = (3, 3)

actions = ['U', 'D', 'L', 'R']
Q = np.zeros((4, 4, 4))

alpha = 0.7
gamma = 0.9
epsilon = 0.9

# Move the agent
def step(state, action):
    r, c = state

    if action == 'U':
        r -= 1
    elif action == 'D':
        r += 1
    elif action == 'L':
        c -= 1
    elif action == 'R':
        c += 1

    if r < 0 or r >= 4 or c < 0 or c >= 4 or maze[r][c] == 1:
        return state, -1

    if (r, c) == goal:
        return (r, c), 10

    return (r, c), -0.1

# Training
for episode in range(500):
    state = start

    for _ in range(100):
        if random.random() < epsilon:
            action = random.choice(actions)
        else:
            action = actions[np.argmax(Q[state[0], state[1]])]

        next_state, reward = step(state, action)
        a = actions.index(action)

        Q[state[0], state[1], a] += alpha * (
            reward + gamma * np.max(Q[next_state[0], next_state[1]])
            - Q[state[0], state[1], a]
        )

        state = next_state

        if state == goal:
            break

# Test the learned path
state = start
path = [state]

for _ in range(100):
    if state == goal:
        break

    action = actions[np.argmax(Q[state[0], state[1]])]
    state, _ = step(state, action)
    path.append(state)

print("Learned Path:")
print(path)
```
